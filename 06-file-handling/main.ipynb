## seek and tell functions

```python
with open('sample.txt', 'r') as f:
    print(f.read(10))
    print(f.tell())
    f.seek(0)
    print(f.read(10))
    print(f.tell())
    
```

## Serializing and deserializing objects
Serialization is the process of converting a Python object into a byte stream or a format (like JSON, pickle, etc.) that can be stored in a file or transmitted over a network.
Deserialization is the reverse process, where the byte stream or stored format is converted back into a Python object.

Example:
- Serialization: Python object -> file/byte stream
- Deserialization: file/byte stream -> Python object

```python
# json is the universal format for serialization and deserialization.
import json
l=[1, 2, 3, 4, 5]
with open('data.json', 'w') as f:
    json.dump(l, f)
```

```python
# decerialization
with open('data.json', 'r') as f:
    data = json.load(f)
    print(data)
    print(type(data))  
    
```

## pickle module

```python
import pickle
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def id(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
```

```python
student1 = Student("Manoj Singh Rawat", 25)
student2 = Student("Rahul Kumar", 23)
with open('student_data.pkl', 'wb') as f:
    pickle.dump(student1, f)  # Save single object
```

## LEGB RULE
LEGB is an acronym that stands for Local, Enclosing, Global, and Built-in. It describes the order in which Python looks for names when you reference a variable or function.
1. **Local**: Names assigned within a function (or lambda) are local to that function.
2. **Enclosing**: Names in the local scope of any enclosing function (i.e., a function defined inside another function).
3. **Global**: Names assigned at the top level of a module or declared global within a function.
4. **Built-in**: Names preassigned in the built-in namespace (like `len`, `print`, etc.).


```python
a=10
def f1():
    global a
    a+=1
    
    print(a)  # This will raise a NameError since 'a' is not defined in this scope

f1()
print(a)  # This will print the updated value of 'a'
```

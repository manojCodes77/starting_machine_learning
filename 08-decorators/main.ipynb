## Decorators

Decorator is a function which receives another function as an input and add some functionality to it and return it

### Types of Decorators
1. **Built-in Decorators**: Python provides several built-in decorators like `@staticmethod`, `@classmethod`, and `@property`.
2. **User-defined Decorators**: You can create your own decorators to add custom functionality.

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Decorator is called")
        print(f"Args: {args}")
        print(f"Kwargs: {kwargs}")
        return func(*args, **kwargs)
    return wrapper
```

```python
def greet(name, age=None, city=None):
    if age and city:
        print(f"Hello {name}, you are {age} years old and live in {city}")
    elif age:
        print(f"Hello {name}, you are {age} years old")
    else:
        print(f"Hello {name}")
```

```python
a=my_decorator(greet)
a("Manoj", city="Uttarakhand")
```

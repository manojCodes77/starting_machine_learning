```python
def step(z):
    return 1 if z>0 else 0
```

```python
def perceptron(x,y):
    x=np.insert(x,0,1,axis=1)
    weights=np.ones(x.shape[1])
    lr=0.1

    for i in range(1000):
        j=np.random.randint(0,100)
        y_hat=step(np.dot(x[j],weights))
        weights=weights+lr*(y[j]-y_hat)*x[j]
    return weights[0],weights[1:]

```

```python
intercept_,coef_=perceptron(x,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m=-coef_[0]/coef_[1]
c=-intercept_/coef_[1]
```

```python
x_input=np.linspace(-3,3,10)
y_input=m*x_input+c
```

```python
plt.figure(figsize=(10,6))
# plt.plot(x_input,y_input,color='red',label='Decision Boundary',linewidth=2)
# i wanna  create a doted line for the decision boundary
plt.plot(x_input,y_input,color='red',label='Decision Boundary',linewidth=2, linestyle='dotted')
plt.scatter(x[:,0],x[:,1],c=y,cmap='Set1',edgecolor='k',s=100,alpha=0.7)
plt.ylim(-3,2)
```

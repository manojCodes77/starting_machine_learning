```python
matrix1=np.random.randint(100, size=(3,4))
matrix2=np.random.randint(100, size=(3,4))
print(matrix1)
print(matrix2)
# element-wise multiplication
result = matrix1 * matrix2
print(result)
# matrix multiplication
result = np.dot(matrix1, matrix2.T)
print(result)

# addition
result = matrix1 + matrix2
print(result)
```

```python
matx=np.arange(1,17).reshape(4,4)
# row wise sum
row_sum = np.sum(matx, axis=1)
print(row_sum)
# column wise sum
col_sum = np.sum(matx, axis=0)
print(col_sum)
```

```python
matrix = np.arange(1, 26).reshape(5, 5)
sd=np.std(matrix)
print(sd)
# mean
mean=np.mean(matrix)
print(mean)
# median
median=np.median(matrix)
print(median)
# variance
variance=np.var(matrix)
print(variance)
```

```python
matrix=np.arange(1,10).reshape(3,3)
# normalization
normalized_matrix = (matrix - np.mean(matrix)) / np.std(matrix)
print(normalized_matrix)
```

```python
# use of broadcasting
matrix1=np.random.randint(1,10, size=(3,3))
array1=np.random.randint(1,10, size=(3,1))
matrix1 = matrix1 + array1
print(matrix1)
```

```python
print("Matrix 1:", matrix1)
temp=np.linalg.det(matrix1)
print("Determinant of Matrix 1:", temp)
# inverse of matrix
temp=np.linalg.inv(matrix1)
print("Inverse of Matrix 1:", temp)
# eigenvalues and eigenvectors
temp=np.linalg.eig(matrix1)
print("Eigenvalues of Matrix 1:", temp[0])
print("Eigenvectors of Matrix 1:", temp[1])
# singular value decomposition
```

```python
import numpy.ma as ma

# Create a masked array of shape (4, 4) with random integers
array = np.random.randint(1, 21, size=(4, 4))
masked_array = ma.masked_greater(array, 10)
print("Original array:")
print(array)
print("Masked array:")
print(masked_array)

# Compute the sum of the unmasked elements
sum_unmasked = masked_array.sum()
print("Sum of unmasked elements:", sum_unmasked)
```

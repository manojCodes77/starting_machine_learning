```python
import seaborn as sns
```

```python
tips=sns.load_dataset("tips")
tips
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(tips["total_bill"], tips["tip"], c=tips["total_bill"], cmap='viridis', alpha=0.5, marker='*')
plt.title("Scatter plot of total bill vs tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.colorbar(label='Total Bill')
plt.show()
```

```python
# line plot
sns.lineplot(x='size', y='total_bill', data=tips, color='purple', alpha=0.5)
plt.title("Line plot of size vs total bill")
plt.xlabel("Size")
plt.ylabel("Total Bill")    
plt.show()
```

```python
# violin plot
sns.violinplot(x='day', y='total_bill', data=tips, palette='pastel')
plt.title("Violin plot of day vs total bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
# pair plot
```

```python
# bins
l=[0,5,10,15,20,25,30,35,40,45,50,55,60]
sns.histplot(tips['total_bill'], bins=l, color='blue', alpha=0.7,kde='Black')
# plt.hist(tips['total_bill'], bins=l, color='blue', alpha=0.7, edgecolor='black', density=True)
# sns.kdeplot(tips['total_bill'], color='red')
# plt.title("Histogram of Total Bill with KDE")
# plt.xlabel("Total Bill")
# plt.ylabel("Density")
# plt.show()
```

```python
# pair plot
plt.figure(figsize=(10, 6))
sns.pairplot(tips, hue='day', palette='pastel')
plt.suptitle("Pair plot of tips dataset", y=1.02)
plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
x=[1, 2, 3, 4, 5]
y1=[2, 3, 5, 7, 11]
y2=[1, 4, 6, 8, 10]
plt.plot(x, y1, marker='o', linestyle='solid', color='b', label='y1')
plt.plot(x, y2, marker='s', linestyle='dotted', color='g', label='y2')
plt.title('Sample Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.savefig('sample_plot.png')
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 5))

axs[0][0].plot(x, y1, marker='o', linestyle='solid', color='b', label='y1')
axs[0][0].set_title('Plot of y1')
axs[0][0].set_xlabel('X-axis')
axs[0][0].set_ylabel('Y-axis')
axs[0][0].grid(True)

axs[1][0].plot(x, y2, marker='s', linestyle='dotted', color='g', label='y2')
axs[1][0].set_title('Plot of y2')
axs[1][0].set_xlabel('X-axis')
axs[1][0].set_ylabel('Y-axis')
axs[1][0].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure()
sc = plt.scatter(x, y, c=y, cmap='magma', label='Sine Wave')
plt.colorbar(sc, label='Color Scale')
plt.title('Sine Wave with Color Map')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
# x = [1, 2, 3, 4, 5]
# y1 = [1, 4, 9, 16, 25]
# y2 = [1, 2, 3, 4, 5]
import numpy as np
x= np.linspace(0, 10, 100)
y1=np.square(x)
y2=x

plt.figure(figsize=(12, 7), dpi=60)

plt.subplot(2,3,1)
plt.plot(x,y1,color='green')
plt.title("Plot 1")

plt.subplot(2, 3, (2, 3))
plt.plot(y1, x, color='red')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot(x,y2,color='blue')
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.plot(x,y2,color='green')
plt.title("Plot 4")
```

```python
# bar plots
import numpy as np
import matplotlib.pyplot as plt
categories=['A', 'B', 'C', 'D','E']
values=[5, 7, 3, 8, 6]
colors = plt.cm.cividis(np.linspace(0, 1, len(values)))
print(colors)
plt.bar(categories, values, color=colors)
```

```python
# histogram
data=[0,1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
# bins parameter
l=[0,1,2,3,4,5]
plt.figure(figsize=(8, 6), dpi=60)
plt.gca().yaxis.set_major_locator(plt.MultipleLocator(1))
plt.hist(data, bins=l, color='purple', alpha=1, edgecolor='black')
plt.grid(True)
plt.show()
```

```python
# scatter plot
x= [1, 2, 3, 4, 5]
y= [2, 3, 5, 7, 11]
plt.scatter(x, y, color='red', marker='x')
plt.title('Scatter Plot')
plt.xlabel('X-axis')
```

```python
# pie chart
labels=['A', 'B', 'C', 'D']
sizes=[15, 30, 45, 10]
colors=['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode=(0.3, 0, 0, 0)  # explode the
# first slice
plt.pie(sizes, explode=explode, labels=labels, colors=colors,
        autopct='%1.1f%%', shadow=True, startangle=90)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('Pie Chart Example')
plt.show()
```

```python
# start playing with csv files
import pandas as pd

df=pd.read_csv('sales-data.csv')
df.head(5)
```

```python
total_sales=df.groupby('Product Category')['Total Revenue'].sum()
print(total_sales)
```

```python
# now plot a bar graph of this
import matplotlib.pyplot as plt

colors = plt.cm.viridis(np.linspace(0, 1, len(total_sales)))
plt.bar(total_sales.index, total_sales.values, color=colors)
plt.title('Total Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45,color='blue')
plt.tight_layout()
plt.show()
```

```python
# iterator
class Countdown:
    def __init__(self, start):
        self.current = start

    def __bhaikanext__(self):
        if self.current <= 0:
            raise StopIteration
        else:
            ans= self.current
            self.current -= 1
            return ans

# Test
countdown = Countdown(5)
# print(countdown)
while True:
    try:
        number = countdown.__bhaikanext__()
        print(number)
    except StopIteration:
        break
```

```python
class MyRange:
    def __init__(self,n):
        self.n = n
        self.current = 0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current >= self.n:
            raise StopIteration
        else:
            ans = self.current
            self.current += 1
            return ans
# Test
my_range = MyRange(5)
for i in my_range:
    print(i)
```

### generator function

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b
# Test
start=fibonacci(11)
while True:
    try:
        number =next(start)
        print(number,end=" ")
    except StopIteration:
        break
```

```python
def square():
    for i in range(1,11):
        yield i**2
res= square()
while True:
    try:
        number =next(res)
        print(number,end=" ")
    except StopIteration:
        break
```

```python
# square of even numbers
def square(i):
    yield i**2
def even_square():
    for i in range(1,21):
        if i%2==0:
            yield square(i)
res= even_square()
while True:
    try:
        number =next(next(res))
        print(number,end=" ")
    except StopIteration:
        break
```

### decorators


```python
import time

def time_it(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time} seconds")
        return result
    return wrapper

@time_it
def factorial(n):
    res=1
    for i in range(1,n+1):
        res *= i
    return res

# Test
print(factorial(10))
```

the above code can also be written as:

```python
def factorial(n):
    res=1
    for i in range(1,n+1):
        res *= i
    return res

factorial=time_it(factorial)

print(factorial(10))
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)  # Pass only the number of repetitions here
def greet(name):
    print(f"Hello, {name}!")

# Test
greet("Alice")
```

```python
a=int(input("Enter a number: "))
b=int(input("Enter another number: "))
try:
    result = a / b
    print("The result is:", result)
    
except ZeroDivisionError:
    print("You cannot divide by zero.")

finally:
    print("Execution completed.")    
```

```python
def read_file(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            print(content)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except IOError:
        print(f"Error: An I/O error occurred while reading the file '{filename}'.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("Finished attempting to read the file.")

read_file("data.txt")
```

```python
lst = [1, 2, 3,-10]

def sum_list(lst):
    total = 0
    for i in lst:
        total += i
    
    try:
        if len(lst) > 0 and total > 0:
            return total
        else:
            if len(lst) == 0:
                raise ValueError("List is empty.")
            else:
                raise ValueError("Sum is not positive.")
    except ValueError as e:
        print(f"ValueError: {e}")
    finally:
        print("Finished calculating the sum.")

print("The sum of the list is:", sum_list(lst))
```

```python
strings= ["hello", "world", 123]
def inserting(s,filename):
    try:
        with open(filename,'w') as file:
            for i in s:
                if isinstance(i, str):
                    file.write(i + "\n")
                else:
                    raise IOError(f"Expected a string but got {type(i).__name__}")
    except IOError as e:
        print(f"Error: An I/O error occurred while writing to the file '{filename}'.\n{e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("Finished attempting to write to the file.")
inserting(strings, "output.txt")
```

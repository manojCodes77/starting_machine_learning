### sqlite is by default comes with python3

```python
import sqlite3
```

```python
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
# create a table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Employee(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER,
    department TEXT NOT NULL
    )
''')

connection.commit()
```

```python
# inserting into the table
cursor.executemany('''
INSERT INTO Employee(name, age, department)
VALUES(?, ?, ?)
''', [
    ('Amit Sharma', 28, 'IT'),
    ('Priya Singh', 32, 'Finance'),
    ('Rajesh Kumar', 45, 'Operations'),
    ('Anita Desai', 29, 'Marketing'),
    ('Vikram Patel', 38, 'HR')
])

connection.commit()
```

```python
cursor.execute('select * from Employee')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

```python
connection.close()
```

## Note on Logging Configuration

Once the basic configuration of logging is set using `logging.basicConfig()`, any subsequent calls to `basicConfig()` will have no effect unless you restart the kernel. To apply new logging configurations, you must restart your kernel and run the configuration code again.

```python
# restart kernel to apply new configuration
import logging

logging.basicConfig(
    format='%(asctime)s - %(levelname)s - %(message)s',
    level=logging.DEBUG,
    filename='app.log',
    filemode='w'
)
```
